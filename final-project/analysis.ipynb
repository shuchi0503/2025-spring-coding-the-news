MANHATTAN

```python
## this is the filtered df for manhattan ems 
man_ems_df= df.query("(`Month Name` == '2025 / 03' or `Month Name` == '2025 / 02' or `Month Name` == '2025 / 01') and Borough == 'Manhattan' \
        and Agency == 'EMS'").sort_values(by="Description")
man_ems_df
```

```python
## just checking frst three rows
df.query("(`Month Name` == '2025 / 03' or `Month Name` == '2025 / 02' or `Month Name` == '2025 / 01') and Borough == 'Manhattan' \
        and Agency == 'EMS'").sort_values(by="Description").head(3)
```

```python
## final response times for manhattan ems!
## dividing final by 60 to get minutes instead of seconds. REMEMBER- to check percentage- remove /60  
man_ems_times_df = man_ems_df.groupby(['Description', 'Month Name'])['Response Times'].mean().unstack()
man_ems_times_df / 60
```

```python
## Now do the same for manhattan fdny and manhattan aggregate 
```

```python
man_fdny_df= df.query("(`Month Name` == '2025 / 03' or `Month Name` == '2025 / 02' or `Month Name` == '2025 / 01') and Borough == 'Manhattan' \
        and Agency == 'FDNY'").sort_values(by="Description")
man_fdny_df
```

```python
## final response times for manhattan fdny
man_fdny_times_df = man_fdny_df.groupby(['Description', 'Month Name'])['Response Times'].mean().unstack()
man_fdny_times_df /60
```
